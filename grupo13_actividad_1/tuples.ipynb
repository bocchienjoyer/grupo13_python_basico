# **1. Imprimir tupla**

```python
ages = (20, 25, 22, 30, 28)

print(ages)
```

# **2. Mostrar el primer y último elemento**

```python
print("Primer elemento:", ages[0])
print("Último elemento:", ages[-1])
```

# **3. Mostrar máximo, mínimo y número de elemento**

```python
print("Valor máximo:", max(ages))
print("Valor mínimo:", min(ages))
print("Número de elementos:", len(ages))
```

# **4. Intentar modificar un elemento**

```python
ages[0] = 21
```
